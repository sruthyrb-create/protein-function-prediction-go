```python
import numpy as np
from scipy import sparse


y_sparse = sparse.load_npz("artifacts/y_sparse.npz")
Y_dense=np.load("artifacts/Y.npy",allow_pickle=False)
print("Shape of sparse matrix:", y_sparse.shape)
print("Type:", type(y_sparse))
print("Density:", y_sparse.nnz / (y_sparse.shape[0] * y_sparse.shape[1]))
```

```python
# Convert a small chunk of the sparse matrix to dense
Y_sparse_chunk = y_sparse[:100, :100].toarray()
Y_dense_chunk = Y_dense[:100, :100]

print("All values identical:", np.allclose(Y_sparse_chunk, Y_dense_chunk))
```

```python
# Number of labels per protein (non-zero elements per row)
labels_per_protein = (y_sparse != 0).sum(axis=1).A1  # .A1 converts to 1D array
print("Average labels per protein:", labels_per_protein.mean())
print("Max labels per protein:", labels_per_protein.max())
```

```python
# Total number of proteins and labels
n_proteins, n_labels = y_sparse.shape
print(f"Proteins: {n_proteins:,}, Labels: {n_labels:,}")

# % of labels that are positive
positive_ratio = y_sparse.nnz / (n_proteins * n_labels) * 100
print(f"Label density: {positive_ratio:.4f}%")

# Distribution summary
label_counts = np.diff(y_sparse.indptr)
print(f"Median labels per protein: {np.median(label_counts)}")
```
